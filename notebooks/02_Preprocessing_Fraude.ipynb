```python
# ============================================
# IMPORTATION DES LIBRAIRIES
# ============================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

from imblearn.over_sampling import SMOTE

import plotly.express as px

print("Librairies importées avec succès")
```

```python
# ============================================
# ANALYSE DES VALEURS MANQUANTES
# ============================================

missing_values = (
    df.isnull()
      .sum()
      .reset_index()
)

missing_values.columns = [
    "Variable",
    "Valeurs_manquantes"
]

missing_values["Pourcentage"] = (
    missing_values["Valeurs_manquantes"] / len(df)
) * 100

missing_values
```

```python
# ============================================
# ENCODAGE DES VARIABLES CATÉGORIELLES
# ============================================

categorical_columns = [
    "type",
    "nameOrig",
    "nameDest"
]

encoder = LabelEncoder()

for col in categorical_columns:
    df[col] = encoder.fit_transform(df[col])

print("Encodage terminé")

df.head()
```

```python
# ============================================
# NORMALISATION DES VARIABLES NUMÉRIQUES
# ============================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

numeric_columns.remove("isFraud")

scaler = StandardScaler()

df[numeric_columns] = scaler.fit_transform(
    df[numeric_columns]
)

print("Normalisation terminée")

df.head()
```

```python
# ============================================
# SÉPARATION FEATURES / TARGET
# ============================================

X = df.drop("isFraud", axis=1)

y = df["isFraud"]

print(X.shape)
print(y.shape)
```

```python
# ============================================
# DISTRIBUTION DES CLASSES AVANT SMOTE
# ============================================

before_smote = (
    y.value_counts()
     .reset_index()
)

before_smote.columns = [
    "Classe",
    "Nombre"
]

before_smote["Classe"] = before_smote["Classe"].map({
    0: "Normale",
    1: "Fraude"
})

fig = px.bar(
    before_smote,
    x="Classe",
    y="Nombre",
    color="Classe",
    text="Nombre",

    color_discrete_map={
        "Normale": "#06B6D4",
        "Fraude": "#7C3AED"
    },

    title="Distribution des classes avant SMOTE"
)

fig.update_layout(
    template="plotly_white",
    height=500,
    title_x=0.5
)

fig.show()
```

```python
# ============================================
# APPLICATION DE SMOTE
# ============================================

smote = SMOTE(random_state=42)

X_resampled, y_resampled = smote.fit_resample(X, y)

print("SMOTE appliqué avec succès")

print(X_resampled.shape)
print(y_resampled.shape)
```

```python
# ============================================
# DISTRIBUTION APRÈS SMOTE
# ============================================

after_smote = (
    pd.Series(y_resampled)
      .value_counts()
      .reset_index()
)

after_smote.columns = [
    "Classe",
    "Nombre"
]

after_smote["Classe"] = after_smote["Classe"].map({
    0: "Normale",
    1: "Fraude"
})

fig = px.bar(
    after_smote,
    x="Classe",
    y="Nombre",
    color="Classe",
    text="Nombre",

    color_discrete_map={
        "Normale": "#06B6D4",
        "Fraude": "#7C3AED"
    },

    title="Distribution des classes après SMOTE"
)

fig.update_layout(
    template="plotly_white",
    height=500,
    title_x=0.5
)

fig.show()
```
