```python
# ============================================
# IMPORTATION DES LIBRAIRIES
# ============================================

import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

print("Librairies importées avec succès")
```

```python
# ============================================
# CHARGEMENT DES DONNÉES
# ============================================

DATA_PATH = "../data/raw/data_cluster.csv"

df = pd.read_csv(DATA_PATH, sep=";")

print("Dataset chargé avec succès")

print(df.shape)

df.head()
```

```python
# ============================================
# INFORMATIONS GÉNÉRALES
# ============================================

df.info()
```

```python
# ============================================
# STATISTIQUES DESCRIPTIVES
# ============================================

df.describe()
```

```python
# ============================================
# ANALYSE DES VALEURS MANQUANTES
# ============================================

missing_data = (
    df.isnull()
      .sum()
      .reset_index()
)

missing_data.columns = ["Variable", "Valeurs_manquantes"]

missing_data["Pourcentage"] = (
    missing_data["Valeurs_manquantes"] / len(df)
) * 100

fig = px.bar(
    missing_data,
    x="Variable",
    y="Pourcentage",
    color="Pourcentage",
    text="Pourcentage",
    color_continuous_scale=["#06B6D4", "#7C3AED", "#0F172A"],
    title="Analyse interactive des valeurs manquantes"
)

fig.update_traces(
    texttemplate='%{text:.2f}%',
    textposition='outside'
)

fig.update_layout(
    template="plotly_white",
    height=550,
    title_x=0.5,
    coloraxis_showscale=False
)

fig.show()
```

```python
# ============================================
# CRÉATION DE LA VARIABLE ÂGE
# ============================================

CURRENT_YEAR = 2026

df["Age"] = CURRENT_YEAR - df["Year_Birth"]

df[["Year_Birth", "Age"]].head()
```

```python
# ============================================
# DISTRIBUTION DES ÂGES
# ============================================

fig = px.histogram(
    df,
    x="Age",
    nbins=30,
    marginal="box",
    color_discrete_sequence=["#06B6D4"],
    title="Distribution des âges des clients"
)

fig.update_layout(
    template="plotly_white",
    height=650,
    title_x=0.5
)

fig.show()
```

```python
# ============================================
# NIVEAU D'ÉDUCATION
# ============================================

education_counts = (
    df["Education"]
    .value_counts()
    .reset_index()
)

education_counts.columns = ["Education", "Nombre"]

fig = px.bar(
    education_counts,
    x="Education",
    y="Nombre",
    color="Nombre",
    text="Nombre",
    color_continuous_scale=["#06B6D4", "#7C3AED", "#0F172A"],
    title="Répartition des niveaux d’éducation"
)

fig.update_traces(textposition='outside')

fig.update_layout(
    template="plotly_white",
    height=600,
    title_x=0.5,
    coloraxis_showscale=False
)

fig.show()
```

```python
# ============================================
# SITUATION MATRIMONIALE
# ============================================

marital_counts = (
    df["Marital_Status"]
    .value_counts()
    .reset_index()
)

marital_counts.columns = ["Situation", "Nombre"]

fig = px.pie(
    marital_counts,
    names="Situation",
    values="Nombre",
    hole=0.55,
    color_discrete_sequence=[
        "#06B6D4",
        "#7C3AED",
        "#0F172A",
        "#94A3B8",
        "#CBD5E1"
    ],
    title="Situation matrimoniale des clients"
)

fig.update_layout(
    template="plotly_white",
    height=650,
    title_x=0.5
)

fig.show()
```

```python
# ============================================
# STRUCTURE FAMILIALE
# ============================================

family_data = pd.DataFrame({
    "Catégorie": ["Enfants", "Adolescents"],
    "Moyenne": [
        df["Kidhome"].mean(),
        df["Teenhome"].mean()
    ]
})

fig = px.bar(
    family_data,
    x="Catégorie",
    y="Moyenne",
    color="Moyenne",
    text="Moyenne",
    color_continuous_scale=["#06B6D4", "#7C3AED"],
    title="Structure familiale moyenne des clients"
)

fig.update_traces(
    texttemplate='%{text:.2f}',
    textposition='outside'
)

fig.update_layout(
    template="plotly_white",
    height=550,
    title_x=0.5,
    coloraxis_showscale=False
)

fig.show()
```

```python
# ============================================
# DISTRIBUTION DES REVENUS
# ============================================

fig = px.histogram(
    df,
    x="Income",
    nbins=50,
    marginal="box",
    color_discrete_sequence=["#7C3AED"],
    title="Distribution des revenus des clients"
)

fig.update_layout(
    template="plotly_white",
    height=650,
    title_x=0.5
)

fig.show()
```

```python
# ============================================
# REVENU MOYEN PAR NIVEAU D'ÉDUCATION
# ============================================

income_by_education = (
    df.groupby("Education")["Income"]
      .mean()
      .reset_index()
      .sort_values(by="Income", ascending=False)
)

fig = px.bar(
    income_by_education,
    x="Education",
    y="Income",
    color="Income",
    text="Income",
    color_continuous_scale=["#06B6D4", "#7C3AED", "#0F172A"],
    title="Revenu moyen par niveau d’éducation"
)

fig.update_traces(
    texttemplate='%{text:.0f}',
    textposition='outside'
)

fig.update_layout(
    template="plotly_white",
    height=600,
    title_x=0.5,
    coloraxis_showscale=False
)

fig.show()
```

```python
# ============================================
# DÉPENSES PAR CATÉGORIE
# ============================================

spending_columns = [
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

spending_means = (
    df[spending_columns]
    .mean()
    .reset_index()
)

spending_means.columns = ["Catégorie", "Dépense_Moyenne"]

fig = px.bar(
    spending_means,
    x="Catégorie",
    y="Dépense_Moyenne",
    color="Dépense_Moyenne",
    text="Dépense_Moyenne",
    color_continuous_scale=["#06B6D4", "#7C3AED", "#0F172A"],
    title="Dépenses moyennes par catégorie de produits"
)

fig.update_traces(
    texttemplate='%{text:.2f}',
    textposition='outside'
)

fig.update_layout(
    template="plotly_white",
    height=650,
    title_x=0.5,
    coloraxis_showscale=False
)

fig.show()
```

```python
# ============================================
# DÉPENSE TOTALE PAR CLIENT
# ============================================

df["Total_Spending"] = df[spending_columns].sum(axis=1)

fig = px.histogram(
    df,
    x="Total_Spending",
    nbins=50,
    marginal="box",
    color_discrete_sequence=["#0F172A"],
    title="Distribution des dépenses totales par client"
)

fig.update_layout(
    template="plotly_white",
    height=650,
    title_x=0.5
)

fig.show()
```

```python
# ============================================
# RELATION REVENU / DÉPENSE TOTALE
# ============================================

fig = px.scatter(
    df,
    x="Income",
    y="Total_Spending",
    color="Education",
    size="Total_Spending",
    hover_data=["Age", "Marital_Status"],
    title="Relation entre revenu et dépense totale"
)

fig.update_layout(
    template="plotly_white",
    height=700,
    title_x=0.5
)

fig.show()
```

```python
# ============================================
# ANALYSE DES CANAUX D'ACHAT
# ============================================

channel_columns = [
    "NumDealsPurchases",
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases"
]

channel_means = (
    df[channel_columns]
    .mean()
    .reset_index()
)

channel_means.columns = ["Canal", "Moyenne"]

fig = px.bar(
    channel_means,
    x="Canal",
    y="Moyenne",
    color="Moyenne",
    text="Moyenne",
    color_continuous_scale=["#06B6D4", "#7C3AED", "#0F172A"],
    title="Utilisation moyenne des canaux d’achat"
)

fig.update_traces(
    texttemplate='%{text:.2f}',
    textposition='outside'
)

fig.update_layout(
    template="plotly_white",
    height=600,
    title_x=0.5,
    coloraxis_showscale=False
)

fig.show()
```

```python
# ============================================
# MATRICE DE CORRÉLATION 
# ============================================

numeric_df = df.select_dtypes(include=np.number)

corr = numeric_df.corr()

fig = px.imshow(
    corr,
    text_auto=".2f",
    color_continuous_scale=["#06B6D4", "#F8FAFC", "#7C3AED"],
    aspect="auto",
    title="Matrice de corrélation des variables clients"
)

fig.update_layout(
    template="plotly_white",
    height=850,
    title_x=0.5
)

fig.show()
```
