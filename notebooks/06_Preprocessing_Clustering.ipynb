```python
# ============================================
# IMPORTATION DES LIBRAIRIES
# ============================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

from sklearn.decomposition import PCA

import plotly.express as px

print("Librairies importées avec succès")
```

```python
# ============================================
# CHARGEMENT DES DONNÉES
# ============================================

DATA_PATH = "../data/raw/data_cluster.csv"

df = pd.read_csv(DATA_PATH, sep=";")

print("Dataset chargé")

print(df.shape)

df.head()
```

. Gestion des valeurs manquantes
</h3>

</div>

```python
# ============================================
# VALEURS MANQUANTES
# ============================================

missing_values = (
    df.isnull()
      .sum()
      .reset_index()
)

missing_values.columns = [
    "Variable",
    "Valeurs_manquantes"
]

missing_values["Pourcentage"] = (
    missing_values["Valeurs_manquantes"] / len(df)
) * 100

missing_values
```

```python
# ============================================
# VISUALISATION DES VALEURS MANQUANTES
# ============================================

fig = px.bar(
    missing_values,

    x="Variable",
    y="Pourcentage",

    color="Pourcentage",

    text="Pourcentage",

    color_continuous_scale=[
        "#06B6D4",
        "#7C3AED",
        "#0F172A"
    ],

    title="Analyse des valeurs manquantes"
)

fig.update_traces(
    texttemplate='%{text:.2f}%',
    textposition='outside'
)

fig.update_layout(
    template="plotly_white",

    height=550,

    title_x=0.5,

    coloraxis_showscale=False
)

fig.show()
```

```python
# ============================================
# IMPUTATION DES VALEURS MANQUANTES
# ============================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:

    df[col] = df[col].fillna(
        df[col].median()
    )

print("Valeurs manquantes traitées")
```

```python
# ============================================
# ENCODAGE DES VARIABLES
# ============================================

categorical_columns = [
    "Education",
    "Marital_Status"
]

encoder = LabelEncoder()

for col in categorical_columns:

    df[col] = encoder.fit_transform(
        df[col]
    )

print("Encodage terminé")

df.head()
```

```python
# ============================================
# TRAITEMENT DES DATES AVANT NORMALISATION
# ============================================

# Conversion de la date client
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    dayfirst=True,
    errors="coerce"
)

# Création d'une variable numérique : ancienneté client
CURRENT_DATE = pd.to_datetime("2026-05-12")

df["Customer_Tenure_Days"] = (
    CURRENT_DATE - df["Dt_Customer"]
).dt.days

# Suppression de la colonne date originale
df = df.drop("Dt_Customer", axis=1)

print("Date transformée en ancienneté client")
```

```python
# ============================================
# NORMALISATION
# ============================================

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print("Normalisation terminée")

scaled_df.head()
```

```python
# ============================================
# PCA
# ============================================

pca = PCA(n_components=2)

pca_components = pca.fit_transform(
    scaled_df
)

pca_df = pd.DataFrame(
    pca_components,
    columns=["PCA1", "PCA2"]
)

print("PCA appliqué avec succès")

pca_df.head()
```

```python
# ============================================
# VISUALISATION PCA
# ============================================

fig = px.scatter(
    pca_df,

    x="PCA1",
    y="PCA2",

    color="PCA1",

    color_continuous_scale=[
        "#06B6D4",
        "#7C3AED",
        "#0F172A"
    ],

    title="Projection PCA des clients"
)

fig.update_layout(
    template="plotly_white",

    height=700,

    title_x=0.5,

    coloraxis_showscale=False
)

fig.show()
```
