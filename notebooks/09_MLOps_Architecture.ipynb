````python
# ============================================
# VALIDATION DES DONNÉES
# ============================================

def validate_dataset(df, required_columns, dataset_name):
    """
    Vérifie que le dataset contient les colonnes obligatoires
    et ne contient pas de doublons critiques.
    """

    print(f"Validation du dataset : {dataset_name}")

    missing_columns = [
        col for col in required_columns
        if col not in df.columns
    ]

    if len(missing_columns) > 0:
        print("Colonnes manquantes :", missing_columns)
    else:
        print("Toutes les colonnes obligatoires sont présentes.")

    print("Nombre de lignes :", df.shape[0])
    print("Nombre de colonnes :", df.shape[1])
    print("Nombre de doublons :", df.duplicated().sum())
    print("Valeurs manquantes totales :", df.isnull().sum().sum())

    print("-" * 60)
````

````python
# ============================================
# COLONNES OBLIGATOIRES — FRAUDE
# ============================================

fraud_required_columns = [
    "step",
    "type",
    "amount",
    "nameOrig",
    "oldbalanceOrg",
    "newbalanceOrig",
    "nameDest",
    "oldbalanceDest",
    "newbalanceDest",
    "isFraud",
    "isFlaggedFraud"
]

validate_dataset(
    fraud_df,
    fraud_required_columns,
    "Détection de fraude"
)
````

````python
# ============================================
# COLONNES OBLIGATOIRES — CLUSTERING
# ============================================

cluster_required_columns = [
    "Year_Birth",
    "Education",
    "Marital_Status",
    "Income",
    "Kidhome",
    "Teenhome",
    "Recency",
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds",
    "NumDealsPurchases",
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases",
    "NumWebVisitsMonth",
    "Response",
    "Complain"
]

validate_dataset(
    cluster_df,
    cluster_required_columns,
    "Segmentation client"
)
````

. Nettoyage automatique
</h3>

````python
# ============================================
# NETTOYAGE AUTOMATIQUE DES DONNÉES
# ============================================

def clean_dataset(df):
    """
    Nettoyage automatique :
    - suppression des doublons ;
    - imputation des valeurs numériques par la médiane ;
    - imputation des valeurs catégorielles par le mode.
    """

    df_clean = df.copy()

    # Suppression des doublons
    df_clean = df_clean.drop_duplicates()

    # Colonnes numériques
    numeric_cols = df_clean.select_dtypes(
        include=np.number
    ).columns

    # Colonnes catégorielles
    categorical_cols = df_clean.select_dtypes(
        include="object"
    ).columns

    # Imputation numérique
    for col in numeric_cols:
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].median()
        )

    # Imputation catégorielle
    for col in categorical_cols:
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].mode()[0]
        )

    return df_clean
````

````python
# ============================================
# APPLICATION DU NETTOYAGE
# ============================================

fraud_clean = clean_dataset(fraud_df)
cluster_clean = clean_dataset(cluster_df)

print("Fraude après nettoyage :", fraud_clean.shape)
print("Clustering après nettoyage :", cluster_clean.shape)

print("Valeurs manquantes fraude :", fraud_clean.isnull().sum().sum())
print("Valeurs manquantes clustering :", cluster_clean.isnull().sum().sum())
````

````python
# ============================================
# SAUVEGARDE DES DONNÉES NETTOYÉES
# ============================================

PROCESSED_DATA_DIR = Path("../data/processed")
PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)

fraud_clean.to_csv(
    PROCESSED_DATA_DIR / "fraud_clean.csv",
    index=False
)

cluster_clean.to_csv(
    PROCESSED_DATA_DIR / "cluster_clean.csv",
    index=False
)

print("Données nettoyées sauvegardées avec succès")
````

````python
# ============================================
# LIBRAIRIES POUR LE VERSIONING
# ============================================

import json
import joblib
from datetime import datetime
from pathlib import Path

print("Librairies versioning importées")
````

````python
# ============================================
# CRÉATION DES DOSSIERS DE VERSIONING
# ============================================

VERSION_DIR = Path("../mlops/versioning")
MODEL_DIR = Path("../models")
PARAMS_DIR = Path("../mlops/params")

VERSION_DIR.mkdir(parents=True, exist_ok=True)
MODEL_DIR.mkdir(parents=True, exist_ok=True)
PARAMS_DIR.mkdir(parents=True, exist_ok=True)

print("Dossiers de versioning créés")
````

````python
# ============================================
# VERSIONING DES DONNÉES
# ============================================

data_version = {
    "version": "v1.0",
    "date_creation": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    "raw_datasets": {
        "fraude": "../data/raw/detection_fraude.csv",
        "clustering": "../data/raw/data_cluster.csv"
    },
    "processed_datasets": {
        "fraud_clean": "../data/processed/fraud_clean.csv",
        "cluster_clean": "../data/processed/cluster_clean.csv",
        "X_resampled": "../data/processed/X_resampled.csv",
        "y_resampled": "../data/processed/y_resampled.csv",
        "cluster_scaled": "../data/processed/cluster_scaled.csv",
        "cluster_pca": "../data/processed/cluster_pca.csv",
        "client_segments_final": "../data/processed/client_segments_final.csv"
    },
    "description": "Version initiale des données utilisées pour la détection de fraude et la segmentation client."
}

with open(VERSION_DIR / "data_version_v1.json", "w", encoding="utf-8") as f:
    json.dump(data_version, f, indent=4, ensure_ascii=False)

print("Version des données sauvegardée")
````

````python
# ============================================
# VERSIONING DES PARAMÈTRES
# ============================================

model_params = {
    "fraud_detection": {
        "problem_type": "classification supervisée",
        "target": "isFraud",
        "models_tested": [
            "LogisticRegression",
            "RandomForestClassifier",
            "XGBClassifier",
            "LGBMClassifier",
            "MLPClassifier"
        ],
        "resampling": "SMOTE",
        "test_size": 0.2,
        "random_state": 42,
        "main_metrics": [
            "Accuracy",
            "Precision",
            "Recall",
            "F1-Score",
            "ROC-AUC"
        ]
    },
    "customer_segmentation": {
        "problem_type": "clustering non supervisé",
        "models_tested": [
            "KMeans",
            "DBSCAN",
            "AgglomerativeClustering",
            "GaussianMixture"
        ],
        "selected_number_of_clusters": 5,
        "selection_method": "Elbow Method",
        "scaling": "StandardScaler",
        "dimensionality_reduction": "PCA",
        "main_metrics": [
            "Silhouette Score",
            "Davies-Bouldin Score"
        ]
    }
}

with open(PARAMS_DIR / "model_params_v1.json", "w", encoding="utf-8") as f:
    json.dump(model_params, f, indent=4, ensure_ascii=False)

print("Paramètres sauvegardés")
````

````python
# ============================================
# EXEMPLE DE VERSIONING DES MODÈLES
# ============================================

model_registry = {
    "version": "v1.0",
    "date_creation": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    "models": {
        "fraud_detection_best_model": {
            "file": "../models/fraud_best_model_v1.pkl",
            "description": "Modèle final retenu pour la détection de fraude bancaire.",
            "status": "à sauvegarder après sélection du meilleur modèle"
        },
        "customer_segmentation_model": {
            "file": "../models/kmeans_customer_segmentation_v1.pkl",
            "description": "Modèle KMeans utilisé pour segmenter les clients en 5 groupes.",
            "status": "à sauvegarder après entraînement final"
        }
    }
}

with open(VERSION_DIR / "model_registry_v1.json", "w", encoding="utf-8") as f:
    json.dump(model_registry, f, indent=4, ensure_ascii=False)

print("Registre des modèles sauvegardé")
````

````python
# ============================================
# SAUVEGARDE DU MODÈLE KMEANS FINAL
# ============================================

from sklearn.cluster import KMeans

scaled_df = pd.read_csv("../data/processed/cluster_scaled.csv")

kmeans_final = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

kmeans_final.fit(scaled_df)

joblib.dump(
    kmeans_final,
    MODEL_DIR / "kmeans_customer_segmentation_v1.pkl"
)

print("Modèle KMeans sauvegardé")
````

````python
# Exemple simplifié FastAPI

from fastapi import FastAPI
import joblib
import pandas as pd

app = FastAPI(
    title="Nexora Analytics API",
    description="API de prédiction pour fraude bancaire et segmentation client",
    version="1.0"
)

# Chargement du modèle de segmentation
kmeans_model = joblib.load("../models/kmeans_customer_segmentation_v1.pkl")

@app.get("/")
def home():
    return {
        "message": "Bienvenue sur l'API Nexora Analytics"
    }

@app.post("/predict-segment")
def predict_segment(data: dict):
    input_df = pd.DataFrame([data])
    prediction = kmeans_model.predict(input_df)

    return {
        "segment_cluster": int(prediction[0])
    }
````

````python
# ============================================
# SIMULATION DE DÉRIVE DES DONNÉES
# ============================================

cluster_path = Path("../data/processed/client_segments_final.csv")

if cluster_path.exists():
    client_data = pd.read_csv(cluster_path)
else:
    client_data = pd.read_csv("../data/raw/data_cluster.csv", sep=";")

reference_income = client_data["Income"].dropna()

# Simulation de nouvelles données avec légère dérive
new_income = reference_income * np.random.normal(
    loc=1.08,
    scale=0.05,
    size=len(reference_income)
)

drift_df = pd.DataFrame({
    "Dataset": ["Données historiques"] * len(reference_income)
               + ["Nouvelles données simulées"] * len(new_income),
    "Income": list(reference_income) + list(new_income)
})

fig = px.histogram(
    drift_df,
    x="Income",
    color="Dataset",
    marginal="box",
    nbins=50,
    barmode="overlay",
    opacity=0.65,
    title="Détection de dérive des données — Variable Income",
    color_discrete_sequence=[
        "#06B6D4",
        "#7C3AED"
    ]
)

fig.update_layout(
    template="plotly_white",
    height=650,
    title_x=0.5
)

fig.show()
````

````python
# ============================================
# INDICATEUR DE DÉRIVE
# ============================================

reference_mean = reference_income.mean()
new_mean = new_income.mean()

drift_rate = abs(new_mean - reference_mean) / reference_mean * 100

drift_summary = pd.DataFrame({
    "Indicateur": [
        "Moyenne historique Income",
        "Moyenne nouvelles données Income",
        "Taux de dérive (%)"
    ],
    "Valeur": [
        round(reference_mean, 2),
        round(new_mean, 2),
        round(drift_rate, 2)
    ]
})

drift_summary
````
