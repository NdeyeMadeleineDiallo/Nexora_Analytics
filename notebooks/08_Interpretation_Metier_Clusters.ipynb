```python
# ============================================
# IMPORTATION DES LIBRAIRIES
# ============================================

import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

from sklearn.cluster import KMeans

from sklearn.preprocessing import StandardScaler

print("Librairies importées")
```

```python
# ============================================
# CHARGEMENT DES DONNÉES
# ============================================

DATA_PATH = "../data/raw/data_cluster.csv"

df = pd.read_csv(DATA_PATH, sep=";")

print(df.shape)

df.head()
```

```python
# ============================================
# PRÉTRAITEMENT RAPIDE
# ============================================

# Gestion des dates
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    dayfirst=True,
    errors="coerce"
)

CURRENT_DATE = pd.to_datetime("2026-05-12")

df["Customer_Tenure_Days"] = (
    CURRENT_DATE - df["Dt_Customer"]
).dt.days

df = df.drop("Dt_Customer", axis=1)

# Gestion des valeurs manquantes
numeric_cols = df.select_dtypes(
    include=np.number
).columns

for col in numeric_cols:

    df[col] = df[col].fillna(
        df[col].median()
    )

# Encodage
categorical_cols = [
    "Education",
    "Marital_Status"
]

for col in categorical_cols:

    df[col] = df[col].astype("category").cat.codes

# Normalisation
scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print("Prétraitement terminé")
```

</h3>

```python
# ============================================
# KMEANS FINAL
# ============================================

kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(
    scaled_df
)

df["Cluster"] = clusters

print("Clusters générés")
```

```python
# ============================================
# PROFILING DES CLUSTERS
# ============================================

cluster_profile = df.groupby("Cluster")[[
    
    "Income",
    
    "MntWines",
    "MntMeatProducts",
    "MntGoldProds",
    
    "NumWebPurchases",
    "NumStorePurchases",
    
    "NumDealsPurchases",
    
    "Recency"
    
]].mean()

cluster_profile
```

```python
# ============================================
# HEATMAP DES CLUSTERS
# ============================================

fig = px.imshow(
    cluster_profile,

    text_auto=".1f",

    color_continuous_scale=[
        "#06B6D4",
        "#F8FAFC",
        "#7C3AED"
    ],

    aspect="auto",

    title="Analyse comparative des segments clients"
)

fig.update_layout(
    template="plotly_white",

    height=700,

    title_x=0.5
)

fig.show()
```

```python
# ============================================
# RÉPARTITION DES SEGMENTS
# ============================================

segment_counts = (
    df["Segment_Client"]
    .value_counts()
    .reset_index()
)

segment_counts.columns = [
    "Segment",
    "Nombre"
]

fig = px.pie(
    segment_counts,

    names="Segment",
    values="Nombre",

    hole=0.55,

    color_discrete_sequence=[
        "#06B6D4",
        "#7C3AED",
        "#0F172A",
        "#94A3B8",
        "#CBD5E1"
    ],

    title="Répartition des segments clients"
)

fig.update_layout(
    template="plotly_white",

    height=700,

    title_x=0.5
)

fig.show()
```

```python
# ============================================
# RADAR CHART DES SEGMENTS
# ============================================

radar_features = [

    "Income",
    "MntWines",
    "NumWebPurchases",
    "NumDealsPurchases"
]

radar_data = (
    df.groupby("Segment_Client")[radar_features]
    .mean()
)

fig = go.Figure()

for segment in radar_data.index:

    fig.add_trace(go.Scatterpolar(

        r=radar_data.loc[segment].values,

        theta=radar_features,

        fill='toself',

        name=segment
    ))

fig.update_layout(

    polar=dict(
        radialaxis=dict(
            visible=True
        )
    ),

    template="plotly_white",

    height=750,

    title="Comparaison des profils clients"
)

fig.show()
```
